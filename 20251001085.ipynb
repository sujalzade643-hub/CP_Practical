```python
import math

raw_kc = input("Enter Kc: ")

print("Raw input:", raw_kc)
print("Raw input data type:", type(raw_kc).__name__)

try:
    kc = float(raw_kc)

    print("Converted value:", kc)
    print("Converted value data type:", type(kc).__name__)

    if not math.isfinite(kc) or kc <= 0:
        print("Error: Kc must be a positive number.")

    else:
        # kc*x^2 - (2.5*kc + 1)*x + kc = 0
        a = kc
        b = -(2.5 * kc + 1)
        c = kc

        discriminant = b**2 - 4 * a * c

        print(f"a = {a}")
        print(f"b = {b}")
        print(f"c = {c}")
        print(f"Discriminant = {discriminant}")

        root1 = (-b + math.sqrt(discriminant)) / (2 * a)
        root2 = (-b - math.sqrt(discriminant)) / (2 * a)

        print(f"Root 1 = {root1}")
        print(f"Root 2 = {root2}")

        if 0 <= root1 <= 1:
            x = root1
        elif 0 <= root2 <= 1:
            x = root2
        else:
            print("Error: No physically meaningful root found.")
            x = None

        if x is not None:
            equilibrium_A = 1.5 - x
            equilibrium_B = 1.0 - x
            equilibrium_C = x

            conversion = x / 1.0 * 100

            print(f"\nSelected root x = {x:.3f}")
            print(f"Equilibrium [A] = {equilibrium_A:.3f} mol/L")
            print(f"Equilibrium [B] = {equilibrium_B:.3f} mol/L")
            print(f"Equilibrium [C] = {equilibrium_C:.3f} mol/L")
            print(f"Conversion of B = {conversion:.3f}%")

            if conversion >= 70:
                classification = "high"
            elif conversion >= 40:
                classification = "moderate"
            else:
                classification = "low"

            print(f"Conversion classification: {classification}")

except ValueError:
```
